# 一.ndarray
## 1.nadarray的特性
### 多维性

```python
import numpy as np
```

```python
arr = np.array(5) #创建0维的ndarray数组
print(arr)
print('arr的维度: ',arr.ndim) #数组的维度
```

```python
arr = np.array([1,2,3]) #创建1维的ndarray数组
print(arr)
print('arr的维度: ',arr.ndim) #数组的维度
```

```python
arr = np.array([[1,2,3],[4,5,6]]) #创建2维的ndarray数组
print(arr)
print('arr的维度: ',arr.ndim) #数组的维度
```

### 同质性

```python
arr = np.array([1,'hello']) #不同的数据类型会被转换成相同的数据类型
print(arr)
```

```python
arr = np.array([1,2.5]) #不同的数据类型会被转换成相同的数据类型
print(arr)
```

## 3.ndarray的创建

```python
# 基础的创建方法
list1 = [4,5,6]
arr = np.array(list1,dtype=np.float64)
print(arr.ndim)
print(arr)
```

```python
# copy
arr1 = np.copy(arr) #通过复制过来的两个数组互不影响
print(arr1)
arr1[0] = 8
print(arr1)
print(arr)
```

```python
#预定义形状
#全0 全1 未初始化 固定值
#全0
arr = np.zeros((2,3)) #里面填的是形状，两行三列
print(arr)
print(arr.dtype) #默认是float类型
```

```python
#全1
arr = np.ones((5,8),dtype=int)
print(arr)
```

```python
#未初始化
arr = np.empty((4,2)) #效率更快，但数值随机
print(arr)
```

```python
#指定除了0和1的数值
arr = np.full((2,3),2026)
print(arr)
```

```python
#模仿形状_like
arr1 = np.ones_like(arr)
print(arr1)
```

```python
#等差数列
arr = np.arange(1,10,1) #start,end,step(步长)
print(arr)
```

```python
#等间隔数列
arr1 = np.linspace(0,100,5,dtype=int) #start,end,取几份
print(arr1)
arr2 = np.arange(0,101,25) #等差数列也能实现
print(arr2)
```

```python
#对数间隔数列
arr = np.logspace(0,4,3,base=2,dtype=int) #先对0到4分成3份--(0,2,4)，再以base为底(默认是10),结果为2的0次方，2的2次方，2的四次方
print(arr)
```

```python
#特殊矩阵
#单位矩阵:主对角线上全为1，其余数字全为0
arr = np.eye(3,dtype=int)
print(arr)
arr = np.eye(3,4,dtype=int) #也可以创建三行四列矩阵，⚠️不用加小括号
print(arr)
```

```python
#对角矩阵:主对角线上数字非零，其他数字为0
arr = np.diag([5,1,2,3]) #括号里面填对角线上的数字
print(arr)
```

```python
#随机数组的生成
#生成0-1之间的随机浮点数(均匀分布)
arr = np.random.rand(2,3) #括号里填形状
print(arr)
```

```python
#生成指定范围区间的随机浮点数
arr = np.random.uniform(3,6,(2,3)) #参数为范围3-6，形状2*3
print(arr)
```

```python
#生成指定范围区间的随机整数
arr = np.random.randint(3,6,(2,3)) #参数为范围3-6，形状2*3
print(arr)
```

```python
# 生成随机数列(正态分布:两边小，中间高)
arr = np.random.randn(2,3)
print(arr)
```

```python
#设置随机种子
np.random.seed(20) #设计种子之后，同一个种子的结果不会变
arr = np.random.randint(1,10,(2,5))
print(arr)
```

## 4.ndarray的数据类型
### 布尔类型bool
###  整数类型int
### 浮点数float
### 复数complex

```python
arr = np.array([1,0,1,0],dtype=np.bool)
print(arr)
arr = np.array([1.0,0,1,0],dtype=int)
print(arr)
```

## 5.ndarray的索引与切片

```python
#一堆数组的索引与切片
arr = np.random.randint(1,100,20)
print(arr)
print(arr[10])
print(arr[:])
print(arr[(arr>10) & (arr<70)]) #布尔索引  
arr1 = np.random.randint(3,6,(2,3))
print(arr1)
print(arr1[1,0:2]) #二维数组切片
```
